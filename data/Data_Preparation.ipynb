# M5 Forecasting Data Preparation

This notebook contains the data preparation and exploration for the M5 Forecasting competition.

## Dataset Overview
The M5 competition dataset contains:
- **calendar.csv**: Calendar information with dates and events
- **sales_train_evaluation.csv**: Training data with evaluation labels (121MB)
- **sales_train_validation.csv**: Training data with validation labels (120MB)
- **sample_submission.csv**: Sample submission format (5MB)
- **sell_prices.csv**: Price information for items (203MB)

## Note
The large CSV files are stored locally but excluded from git due to GitHub's 100MB file size limit.


```python
# Import necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

# Set up plotting
plt.style.use('seaborn-v0_8')
sns.set_palette("husl")

print("Libraries imported successfully!")
```

## Data Loading

Load the dataset files from the data directory.


```python
# Define data directory
data_dir = Path("data")

# Check if data files exist
required_files = [
    "calendar.csv",
    "sales_train_evaluation.csv",
    "sales_train_validation.csv",
    "sample_submission.csv",
    "sell_prices.csv"
]

print("Checking for data files...")
for file in required_files:
    file_path = data_dir / file
    if file_path.exists():
        size_mb = file_path.stat().st_size / (1024 * 1024)
        print(f"✅ {file} ({size_mb:.1f} MB)")
    else:
        print(f"❌ {file} - MISSING")

print("\nNote: Large CSV files are excluded from git due to GitHub's 100MB limit.")
print("Please download them from: https://www.kaggle.com/competitions/m5-forecasting-accuracy/data")
```
